```python
df.columns = df.columns.str.strip()
```

```python
df['Education'] = df['Education'].replace({
    'Basic' : 'Undergraduate',
    'Graduation' : 'Graduate',
    '2n Cycle' : 'Graduate', 
    'Master' : 'Postgraduate',
    'PhD' : 'Postgraduate'
 })
```

```python
df['Education'].value_counts()
```

```python
df['Marital_Status'].value_counts()
```

```python
df['Marital_Status'] = df['Marital_Status'].replace({
    'Married' : 'Partner',
    'Together' : 'Partner', 
    'Single' : 'Alone',
    'Divorced' : 'Alone',
    'Widow' : 'Alone',
    'YOLO' : 'Alone', 
    'Absurd' : 'Alone'
})
```

```python
df['Marital_Status'].value_counts()
```

```python
df['Income'] = df['Income'].replace('[\$,]', '', regex=True)
```

```python
df['Income'] = pd.to_numeric(df['Income'], errors='coerce')
```

```python
df['Income'].dtype
```

```python
df['Income'] = df.groupby(['Education', 'Marital_Status'])['Income'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
df.isnull().sum()
```

```python
df['Age'] = 2026 - df['Year_Birth']
```

```python
df.head()
```

```python
df['Total_Spending'] = df[['MntWines', 'MntFruits' , 'MntMeatProducts', 
                          'MntFishProducts', 'MntSweetProducts', 
                          'MntGoldProds']].sum(axis=1)
```

```python
df['Total_Children'] = df['Kidhome'] + df['Teenhome']
```

```python
df['Total_Purchases'] = df[['NumDealsPurchases', 'NumWebPurchases',
                           'NumCatalogPurchases', 'NumStorePurchases']].sum(axis=1)
```

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
```

```python
df['Customer_Tenure'] = (pd.to_datetime('today') - df['Dt_Customer']).dt.days
```

```python
df['Avg_Spending_per_Purchase'] = df['Total_Spending'] / df['Total_Purchases']
```

```python
df.head()
```

```python
df['Total_Purchases'] = df[['NumWebPurchases', 
                           'NumCatalogPurchases', 
                           'NumStorePurchases']].sum(axis=1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in num_cols:
    plt.figure()
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
def cap_outliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    return column.clip(lower, upper)

num_cols = ['Income', 'Age', 'Total_Spending', 'Total_Purchases']

for col in num_cols:
    df[col] = cap_outliers(df[col])
```

```python
for col in num_cols:
    plt.figure()
    sns.boxplot(x=df[col])
    plt.title(f'After Treatment: {col}')
    plt.show()
```

```python
education_order = {
    'Undergraduate': 0,
    'Graduate': 1,
    'Postgraduate': 2
}

df['Education'] = df['Education'].map(education_order)
```

```python
df['Age_Group'] = pd.cut(df['Age'],
                        bins=[20, 40, 60, 100],
                        labels=['Young', 'Middle', 'Old'])
```

```python
age_purchase = df.groupby('Age_Group')[['NumStorePurchases', 'NumWebPurchases']].mean()
age_purchase
```

```python
age_purchase.plot(kind='bar', figsize=(8,5))
plt.title("Purchase Behavior by Age Group")
plt.ylabel("Average Purchases")
plt.xticks(rotation=0)
plt.show()
```

```python
df['Children_Group'] = df['Total_Children'].apply(
    lambda x: 'With Children' if x > 0 else 'No Children'
)
```

```python
children_purchase = df.groupby('Children_Group')['NumWebPurchases'].mean()
children_purchase
```

```python
children_purchase.plot(kind='bar')
plt.title("Web Purchases: With vs Without Children")
plt.ylabel("Average Web Purchases")
plt.xticks(rotation=0)
plt.show()
```
